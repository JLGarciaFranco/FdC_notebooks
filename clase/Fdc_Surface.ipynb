```python
import numpy as np
```

```python
eta=6.112
epsilon=0.622
```

```python
def qstar(T):
    P=1000
    outer=epsilon*eta/P
    exp_arg=17.3*T/(T+237.3)
    return outer*np.exp(exp_arg)
```

```python
qsstar=qstar(15)
```

```python
qastar=qstar(11.5)
```

```python
qs=qsstar*0.5
qa=qastar*0.6
```

```python
print(qs,qa)
```

```python
qs-qa
```

```python
cp=1005
Lq=2.26e6
def Bo_from_difs(qdif,Tdif):
    return cp*Tdif/(Lq*qdif)
```

```python
Bo_from_difs(qs-qa,3.5)
```
