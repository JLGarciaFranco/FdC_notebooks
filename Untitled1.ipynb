```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable
import matplotlib.cm as cm

# Parameters
U_wind = 30  # wind speed (m/s)
f = 8e-4     # Coriolis parameter (1/s)
nu = 5e-1    # eddy viscosity (m^2/s)

# Depth array
z = np.linspace(0, -100, 600)  # negative downward
d = -z  # depth positive downward

# Analytical Ekman solution
# U(z) = U_0 * exp(d*sqrt(f/(2*nu))) * cos(pi/4 - d*sqrt(f/(2*nu)))
# V(z) = U_0 * exp(d*sqrt(f/(2*nu))) * sin(pi/4 - d*sqrt(f/(2*nu)))

decay = np.sqrt(f / (2 * nu))
U = U_wind * np.exp(-d * decay) * np.cos(np.pi/4 - d * decay)
V = U_wind * np.exp(-d * decay) * np.sin(np.pi/4 - d * decay)

# Normalize colors by depth
norm = Normalize(vmin=0, vmax=100)
colors = cm.viridis(norm(d))

# Create figure with three subplots
fig = plt.figure(figsize=(16, 5))

# Plot 1: U,V versus Depth
ax1 = fig.add_subplot(131)
ax1.plot(U, z, 'b-', linewidth=2, label='U (Eastward)')
ax1.plot(V, z, 'r-', linewidth=2, label='V (Northward)')
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.3)
ax1.set_xlabel('Velocity (m/s)', fontsize=11)
ax1.set_ylabel('Depth (m)', fontsize=11)
ax1.set_title('Ekman Spiral: Velocity Components vs Depth', fontsize=12, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot 2: Hodograph (x-y plane) with depth as color
ax2 = fig.add_subplot(132)
for i in range(len(z)-1):
    ax2.plot(U[i:i+2], V[i:i+2], color=colors[i], linewidth=3)
ax2.scatter(U_wind, 0, color='gold', s=200, marker='>', label='Wind direction', zorder=5, edgecolor='black', linewidth=2)
ax2.scatter(U[0], V[0], color='darkgreen', s=100, marker='o', label='Surface current (z=0)', zorder=5, edgecolor='black', linewidth=1.5)
ax2.scatter(0, 0, color='red', s=100, marker='x', linewidth=2.5, label='No flow', zorder=5)
ax2.set_xlabel('U - Eastward (m/s)', fontsize=11)
ax2.set_ylabel('V - Northward (m/s)', fontsize=11)
ax2.set_title('Hodograph: Current Direction vs Depth', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.axis('equal')
ax2.legend(fontsize=9)

# Add colorbar for depth
sm = ScalarMappable(cmap=cm.viridis, norm=norm)
sm.set_array([])
cbar2 = plt.colorbar(sm, ax=ax2)
cbar2.set_label('Depth (m)', fontsize=10)

# Plot 3: Integrated flow showing 90-degree turning
ax3 = fig.add_subplot(133)

# Numerical integration of velocity field
integrated_u = np.zeros_like(z)
integrated_v = np.zeros_like(z)
integrated_u[0] = U[0]
integrated_v[0] = V[0]

dz = z[1] - z[0]
for i in range(1, len(z)):
    integrated_u[i] = integrated_u[i-1] + U[i] * dz
    integrated_v[i] = integrated_v[i-1] + V[i] * dz

# Plot integrated transport
for i in range(len(z)-1):
    ax3.plot(integrated_u[i:i+2], integrated_v[i:i+2], color=colors[i], linewidth=3)

ax3.scatter(U_wind, 0, color='gold', s=200, marker='>', label='Wind direction', zorder=5, edgecolor='black', linewidth=2)
ax3.scatter(integrated_u[0], integrated_v[0], color='darkgreen', s=100, marker='o', label='Surface', zorder=5, edgecolor='black', linewidth=1.5)
ax3.scatter(integrated_u[-1], integrated_v[-1], color='purple', s=150, marker='s', label='Bottom (100m)', zorder=5, edgecolor='black', linewidth=1.5)
ax3.scatter(0, 0, color='red', s=100, marker='x', linewidth=2.5, zorder=5)

# Add angle annotations
angle_surface = np.degrees(np.arctan2(V[0], U[0]))
angle_integrated = np.degrees(np.arctan2(integrated_v[-1], integrated_u[-1]))

ax3.text(0.05, 0.95, f'Surface angle: {angle_surface:.1f}°\nIntegrated angle: {angle_integrated:.1f}°', 
         transform=ax3.transAxes, fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

ax3.set_xlabel('Integrated U (m·s/m)', fontsize=11)
ax3.set_ylabel('Integrated V (m·s/m)', fontsize=11)
ax3.set_title('Mass Transport: Cumulative Flow Integration', fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3)
ax3.axis('equal')
ax3.legend(fontsize=9)

# Add colorbar for depth
cbar3 = plt.colorbar(sm, ax=ax3)
cbar3.set_label('Depth (m)', fontsize=10)

plt.tight_layout()
plt.show()

# Print some useful info
print(f"Surface current angle from wind: {angle_surface:.2f}°")
print(f"Integrated transport angle from wind: {angle_integrated:.2f}°")
print(f"Surface current magnitude: {np.sqrt(U[0]**2 + V[0]**2):.3f} m/s")
print(f"Total integrated transport magnitude: {np.sqrt(integrated_u[-1]**2 + integrated_v[-1]**2):.3f} m·s/m")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable
import matplotlib.cm as cm

# Parameters
tau = 0.9  # wind stress magnitude (Pa)
rho = 1025  # water density (kg/m^3)
f = 1e-4     # Coriolis parameter (1/s)
nu = 0.6     # eddy viscosity (m^2/s)

# Wind blows northward (V-direction)
# Depth array
z = np.linspace(0, -300, 300)  # negative downward, stop at 300m
d = -z  # depth positive downward

# Analytical Ekman solution
# Classic Ekman spiral: wind stress in y-direction (north)
# u(z) = u_0 * exp(-d*sqrt(f/(2*nu))) * sin(d*sqrt(f/(2*nu)) + pi/4)
# v(z) = u_0 * exp(-d*sqrt(f/(2*nu))) * cos(d*sqrt(f/(2*nu)) + pi/4)
# where u_0 = tau_0 / (rho * sqrt(2) * nu * f)

decay = np.sqrt(f / (2 * nu))
u_0 = tau / (rho * np.sqrt(2) * nu * f)

U = u_0 * np.exp(-d * decay) * np.sin(d * decay + np.pi/4)
V = u_0 * np.exp(-d * decay) * np.cos(d * decay + np.pi/4)

# Normalize colors by depth
norm = Normalize(vmin=0, vmax=300)
colors = cm.viridis(norm(d))

# Create figure with three subplots
fig = plt.figure(figsize=(16, 5))

# Plot 1: U,V versus Depth
ax1 = fig.add_subplot(131)
ax1.plot(U, z, 'b-', linewidth=2, label='U (Eastward)')
ax1.plot(V, z, 'r-', linewidth=2, label='V (Northward)')
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.3)
ax1.set_xlabel('Velocity (m/s)', fontsize=11)
ax1.set_ylabel('Depth (m)', fontsize=11)
ax1.set_title('Ekman Spiral: Velocity Components vs Depth', fontsize=12, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot 2: Hodograph (x-y plane) with discrete vectors and depth as color
ax2 = fig.add_subplot(132)

# Plot continuous spiral first
for i in range(len(z)-1):
    ax2.plot(U[i:i+2], V[i:i+2], color=colors[i], linewidth=1, alpha=0.3)

# Plot discrete vectors every 20m
vector_spacing = 20  # meters
vector_indices = np.where(d % vector_spacing < (z[1] - z[0]))[0]
if len(vector_indices) == 0:
    # Fallback: use index spacing
    vector_indices = np.arange(0, len(z), max(1, int(vector_spacing / np.abs(z[1] - z[0]))))

for idx in vector_indices:
    if idx < len(U):
        arrow_scale = 0.8
        ax2.arrow(0, 0, U[idx]*arrow_scale, V[idx]*arrow_scale, 
                 head_width=0.15, head_length=0.15, fc=colors[idx], ec=colors[idx], 
                 linewidth=2, alpha=0.8, zorder=4)

ax2.arrow(0, 0, 0, 3, head_width=0.2, head_length=0.2, fc='gold', ec='black', linewidth=2, zorder=5, label='Wind (Northward)')
ax2.scatter(U[0], V[0], color='darkgreen', s=100, marker='o', label='Surface current (z=0)', zorder=5, edgecolor='black', linewidth=1.5)
ax2.scatter(0, 0, color='red', s=100, marker='x', linewidth=2.5, label='Origin', zorder=5)
ax2.set_xlabel('U - Eastward (m/s)', fontsize=11)
ax2.set_ylabel('V - Northward (m/s)', fontsize=11)
ax2.set_title('Hodograph: Vector Arrows at Depth Intervals', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.axis('equal')
ax2.legend(fontsize=9)

# Add colorbar for depth
sm = ScalarMappable(cmap=cm.viridis, norm=norm)
sm.set_array([])
cbar2 = plt.colorbar(sm, ax=ax2)
cbar2.set_label('Depth (m)', fontsize=10)

# Plot 3: Integrated flow showing 90-degree turning
ax3 = fig.add_subplot(133)

# Numerical integration of velocity field
integrated_u = np.zeros_like(z)
integrated_v = np.zeros_like(z)
integrated_u[0] = U[0]
integrated_v[0] = V[0]

dz = z[1] - z[0]
for i in range(1, len(z)):
    integrated_u[i] = integrated_u[i-1] + U[i] * dz
    integrated_v[i] = integrated_v[i-1] + V[i] * dz

# Plot integrated transport with discrete vectors
for i in range(len(z)-1):
    ax3.plot(integrated_u[i:i+2], integrated_v[i:i+2], color=colors[i], linewidth=1, alpha=0.3)

# Plot discrete vectors for integrated flow every 20m
for idx in vector_indices:
    if idx < len(integrated_u):
        arrow_scale = 0.3
        ax3.arrow(0, 0, integrated_u[idx]*arrow_scale, integrated_v[idx]*arrow_scale,
                 head_width=0.5, head_length=0.5, fc=colors[idx], ec=colors[idx],
                 linewidth=2, alpha=0.8, zorder=4)

ax3.arrow(0, 0, 0, 30, head_width=2, head_length=2, fc='gold', ec='black', linewidth=2, zorder=5, label='Wind (Northward)')
ax3.scatter(integrated_u[0], integrated_v[0], color='darkgreen', s=100, marker='o', label='Surface', zorder=5, edgecolor='black', linewidth=1.5)
ax3.scatter(integrated_u[-1], integrated_v[-1], color='purple', s=150, marker='s', label='Bottom (100m)', zorder=5, edgecolor='black', linewidth=1.5)
ax3.scatter(0, 0, color='red', s=100, marker='x', linewidth=2.5, zorder=5)

# Add angle annotations
angle_surface = np.degrees(np.arctan2(V[0], U[0]))
angle_integrated = np.degrees(np.arctan2(integrated_v[-1], integrated_u[-1]))

ax3.text(0.05, 0.95, f'Surface angle: {angle_surface:.1f}°\nIntegrated angle: {angle_integrated:.1f}°', 
         transform=ax3.transAxes, fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

ax3.set_xlabel('Integrated U (m·s/m)', fontsize=11)
ax3.set_ylabel('Integrated V (m·s/m)', fontsize=11)
ax3.set_title('Mass Transport: Cumulative Flow Integration', fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3)
ax3.axis('equal')
ax3.legend(fontsize=9)

# Add colorbar for depth
cbar3 = plt.colorbar(sm, ax=ax3)
cbar3.set_label('Depth (m)', fontsize=10)

plt.tight_layout()
plt.show()

# Print some useful info
print(f"Surface current angle from wind: {angle_surface:.2f}°")
print(f"Integrated transport angle from wind: {angle_integrated:.2f}°")
print(f"Surface current magnitude: {np.sqrt(U[0]**2 + V[0]**2):.3f} m/s")
print(f"Total integrated transport magnitude: {np.sqrt(integrated_u[-1]**2 + integrated_v[-1]**2):.3f} m·s/m")
print(f"E-folding depth: {1/decay:.1f} m")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable
import matplotlib.cm as cm

# Parameters
tau = 0.1  # wind stress magnitude (Pa)
rho = 1025  # water density (kg/m^3)
f = 1e-4     # Coriolis parameter (1/s)
nu = 0.1     # eddy viscosity (m^2/s)

# Wind blows northward (V-direction)
# Depth array
z = np.linspace(0, -300, 300)  # negative downward, stop at 300m
d = -z  # depth positive downward

# Analytical Ekman solution
# Classic Ekman spiral: wind stress in y-direction (north)
# u(z) = u_0 * exp(-d*sqrt(f/(2*nu))) * sin(d*sqrt(f/(2*nu)) + pi/4)
# v(z) = u_0 * exp(-d*sqrt(f/(2*nu))) * cos(d*sqrt(f/(2*nu)) + pi/4)
# where u_0 = tau_0 / (rho * sqrt(2) * nu * f)

decay = np.sqrt(f / (2 * nu))
u_0 = tau / (rho * np.sqrt(2) * nu * f)

U = u_0 * np.exp(-d * decay) * np.sin(d * decay + np.pi/4)
V = u_0 * np.exp(-d * decay) * np.cos(d * decay + np.pi/4)

# Normalize colors by depth
norm = Normalize(vmin=0, vmax=500)
colors = cm.viridis(norm(d))

# Create figure with three subplots
fig = plt.figure(figsize=(16, 5))

# Plot 1: U,V versus Depth
ax1 = fig.add_subplot(131)
ax1.plot(U, z, 'b-', linewidth=2, label='U (Eastward)')
ax1.plot(V, z, 'r-', linewidth=2, label='V (Northward)')
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.3)
ax1.set_xlabel('Velocity (m/s)', fontsize=11)
ax1.set_ylabel('Depth (m)', fontsize=11)
ax1.set_title('Ekman Spiral: Velocity Components vs Depth', fontsize=12, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot 2: Hodograph (x-y plane) with discrete vectors and depth as color
ax2 = fig.add_subplot(132)

# Plot continuous spiral first
for i in range(len(z)-1):
    ax2.plot(U[i:i+2], V[i:i+2], color=colors[i], linewidth=1, alpha=0.3)

# Plot discrete vectors every 20m
vector_spacing = 20  # meters
vector_indices = np.where(d % vector_spacing < (z[1] - z[0]))[0]
if len(vector_indices) == 0:
    # Fallback: use index spacing
    vector_indices = np.arange(0, len(z), max(1, int(vector_spacing / np.abs(z[1] - z[0]))))

for idx in vector_indices:
    if idx < len(U):
        arrow_scale = 0.8
        ax2.arrow(0, 0, U[idx]*arrow_scale, V[idx]*arrow_scale, 
                 head_width=0.15, head_length=0.15, fc=colors[idx], ec=colors[idx], 
                 linewidth=2, alpha=0.8, zorder=4)

ax2.arrow(0, 0, 0, 3, head_width=0.2, head_length=0.2, fc='gold', ec='black', linewidth=2, zorder=5, label='Wind (Northward)')
ax2.scatter(U[0], V[0], color='darkgreen', s=100, marker='o', label='Surface current (z=0)', zorder=5, edgecolor='black', linewidth=1.5)
ax2.scatter(0, 0, color='red', s=100, marker='x', linewidth=2.5, label='Origin', zorder=5)
ax2.set_xlabel('U - Eastward (m/s)', fontsize=11)
ax2.set_ylabel('V - Northward (m/s)', fontsize=11)
ax2.set_title('Hodograph: Vector Arrows at Depth Intervals', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.axis('equal')
ax2.legend(fontsize=9)

# Add colorbar for depth
sm = ScalarMappable(cmap=cm.viridis, norm=norm)
sm.set_array([])
cbar2 = plt.colorbar(sm, ax=ax2)
cbar2.set_label('Depth (m)', fontsize=10)

# Plot 3: Integrated flow showing 90-degree turning
ax3 = fig.add_subplot(133)

# Numerical integration of velocity field
integrated_u = np.zeros_like(z)
integrated_v = np.zeros_like(z)
integrated_u[0] = U[0]
integrated_v[0] = V[0]

dz = z[1] - z[0]
for i in range(1, len(z)):
    integrated_u[i] = integrated_u[i-1] + U[i] * dz
    integrated_v[i] = integrated_v[i-1] + V[i] * dz

# Plot integrated transport with discrete vectors
for i in range(len(z)-1):
    ax3.plot(integrated_u[i:i+2], integrated_v[i:i+2], color=colors[i], linewidth=1, alpha=0.3)

# Plot discrete vectors for integrated flow every 20m
for idx in vector_indices:
    if idx < len(integrated_u):
        arrow_scale = 0.3
        ax3.arrow(0, 0, integrated_u[idx]*arrow_scale, integrated_v[idx]*arrow_scale,
                 head_width=0.5, head_length=0.5, fc=colors[idx], ec=colors[idx],
                 linewidth=2, alpha=0.8, zorder=4)

ax3.arrow(0, 0, 0, 30, head_width=2, head_length=2, fc='gold', ec='black', linewidth=2, zorder=5, label='Wind (Northward)')
ax3.scatter(integrated_u[0], integrated_v[0], color='darkgreen', s=100, marker='o', label='Surface', zorder=5, edgecolor='black', linewidth=1.5)
ax3.scatter(integrated_u[-1], integrated_v[-1], color='purple', s=150, marker='s', label='Bottom (100m)', zorder=5, edgecolor='black', linewidth=1.5)
ax3.scatter(0, 0, color='red', s=100, marker='x', linewidth=2.5, zorder=5)

# Add angle annotations
angle_surface = np.degrees(np.arctan2(V[0], U[0]))
angle_integrated = np.degrees(np.arctan2(integrated_v[-1], integrated_u[-1]))

ax3.text(0.05, 0.95, f'Surface angle: {angle_surface:.1f}°\nIntegrated angle: {angle_integrated:.1f}°', 
         transform=ax3.transAxes, fontsize=10, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

ax3.set_xlabel('Integrated U (m·s/m)', fontsize=11)
ax3.set_ylabel('Integrated V (m·s/m)', fontsize=11)
ax3.set_title('Mass Transport: Cumulative Flow Integration', fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3)
ax3.axis('equal')
ax3.legend(fontsize=9)

# Add colorbar for depth
cbar3 = plt.colorbar(sm, ax=ax3)
cbar3.set_label('Depth (m)', fontsize=10)

plt.tight_layout()
plt.show()

# Print some useful info
print(f"Surface current angle from wind: {angle_surface:.2f}°")
print(f"Integrated transport angle from wind: {angle_integrated:.2f}°")
print(f"Surface current magnitude: {np.sqrt(U[0]**2 + V[0]**2):.3f} m/s")
print(f"Total integrated transport magnitude: {np.sqrt(integrated_u[-1]**2 + integrated_v[-1]**2):.3f} m·s/m")
print(f"E-folding depth: {1/decay:.1f} m")
```
