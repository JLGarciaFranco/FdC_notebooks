```python
def M_u_lat(u0,phi1,phi2,a=6.371e6,omega=7.29e-5):
    M1=a*np.cos(np.deg2rad(phi1))*(u0+omega*a*np.cos(np.deg2rad(phi1)))
    print(M1)
    u = M1/(a*np.cos(np.deg2rad(phi2)))-omega*a*np.cos(np.deg2rad(phi2))
    return u
```
